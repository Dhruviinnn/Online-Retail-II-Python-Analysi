```python
sales_df["InvoiceDate"].max()
analysis_date = sales_df["InvoiceDate"].max()
print(f"Latest Transaction Date: {analysis_date}")
```

```python
# RFM Table
rfm = (
    customer_df
    .groupby("Customer ID")
    .agg(
        Recency=("InvoiceDate", lambda x: (analysis_date - x.max()).days),
        Frequency=("Invoice", "nunique"),
        Monetary=("TotalSales", "sum")
    )
    .reset_index()
)
rfm.head()
```

```python
rfm["Recency"].describe()
rfm["Frequency"].describe()
rfm["Monetary"].describe()
```

```python
# RFM Distribution
plt.figure(figsize=(10, 6))

sns.histplot(
    data=rfm,
    x="Recency",
    bins=50,
    kde=True
)

plt.title("Customer Recency Distribution")
plt.xlabel("Recency (Days)")
plt.ylabel("Number of Customers")

plt.tight_layout()

plt.savefig(
    "C:/Users/Vasava Dhruvin/OneDrive/Pictures/Screenshots/RFM_Recency.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=rfm,
    x="Frequency",
    bins=30,
    kde=True
)

plt.title("Customer Purchase Frequency")
plt.xlabel("Number of Orders")
plt.ylabel("Number of Customers")

plt.tight_layout()

plt.savefig(
    "C:/Users/Vasava Dhruvin/OneDrive/Pictures/Screenshots/RFM_Frequency.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=rfm,
    x="Monetary",
    bins=50,
    kde=True
)

plt.title("Customer Monetary Value Distribution")
plt.xlabel("Total Customer Spending")
plt.ylabel("Number of Customers")

plt.tight_layout()

plt.savefig(
    "C:/Users/Vasava Dhruvin/OneDrive/Pictures/Screenshots/RFM_Monetary.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# RFM Scores
rfm["R_Score"] = pd.qcut(
    rfm["Recency"],
    4,
    labels=[4, 3, 2, 1]
)
```

```python
rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
)
```

```python
rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
)
```

```python
rfm["R_Score"] = rfm["R_Score"].astype(int)
rfm["F_Score"] = rfm["F_Score"].astype(int)
rfm["M_Score"] = rfm["M_Score"].astype(int)
rfm.head()
```

```python
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str)
    + rfm["F_Score"].astype(str)
    + rfm["M_Score"].astype(str)
)
```

```python
rfm["RFM_Total"] = (
    rfm["R_Score"]
    + rfm["F_Score"]
    + rfm["M_Score"]
)
```

```python
# Customer Segments
rfm["Segment"] = np.select(
    [
        rfm["RFM_Total"] >= 10,
        rfm["RFM_Total"].between(8, 9),
        rfm["RFM_Total"].between(6, 7),
        rfm["RFM_Total"].between(4, 5)
    ],
    [
        "Champions",
        "Loyal Customers",
        "Potential Loyalists",
        "At Risk"
    ],
    default="Lost Customers"
)
rfm["Segment"].value_counts()
```

```python
segment_counts = (
    rfm["Segment"]
    .value_counts()
    .reset_index()
)

segment_counts.columns = ["Segment", "Customers"]
```

```python
# Customer Retention
customer_df["CohortMonth"] = (
    customer_df
    .groupby("Customer ID")["InvoiceDate"]
    .transform("min")
    .dt.to_period("M")
)
```

```python
customer_df["CohortIndex"] = (
    (customer_df["PurchaseMonth"].dt.year -
     customer_df["CohortMonth"].dt.year) * 12
    +
    (customer_df["PurchaseMonth"].dt.month -
     customer_df["CohortMonth"].dt.month)
)
```

```python
customer_df[
    [
        "Customer ID",
        "PurchaseMonth",
        "CohortMonth",
        "CohortIndex"
    ]
].head(20)
```

```python
# Build Cohort Table
cohort_data = (
    customer_df
    .groupby(
        ["CohortMonth", "CohortIndex"]
    )["Customer ID"]
    .nunique()
    .reset_index()
)

cohort_data.head()
```

```python
cohort_table = cohort_data.pivot(
    index="CohortMonth",
    columns="CohortIndex",
    values="Customer ID"
)
cohort_table.head()
```

```python
# Retention Percentage
cohort_retention = (
    cohort_table
    .divide(
        cohort_table.iloc[:, 0],
        axis=0
    )
)
cohort_retention = cohort_retention * 100
cohort_retention.head()
```

```python
# Retention Heatmap
plt.figure(figsize=(15, 8))

sns.heatmap(
    cohort_retention,
    annot=True,
    fmt=".1f",
    cmap="Blues",
    vmin=0,
    vmax=100
)

plt.title("Customer Retention by Cohort")
plt.xlabel("Months Since First Purchase")
plt.ylabel("Customer Cohort")

plt.tight_layout()

plt.savefig(
    "C:/Users/Vasava Dhruvin/OneDrive/Pictures/Screenshots/CustomerRetentionHeatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Average Retention by Month
average_retention = (
    cohort_retention
    .mean(axis=0)
    .reset_index()
)

average_retention.columns = [
    "Month",
    "RetentionRate"
]

average_retention
```
